```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
k=8
a=np.arange(1,10006,2)
domain=np.linspace(-3.1415,3.1415,1000)
```

```python
function=np.zeros(len(domain))
for i in range (len(domain)):
  for j in a:
    function[i]+=(4*k/(np.pi*j))*np.sin(j*domain[i])
```

```python
plt.plot(domain,function)
plt.show()
```
